# 4. Prepare the dataset

## 4.1 Correct Consolidation on Three Levels

```python
# Create separate DataFrames for each period
df_short = top_tracks_short.copy()
df_short["period"] = "short_term"

df_medium = top_tracks_medium.copy()
df_medium["period"] = "medium_term"

df_long = top_tracks_long.copy()
df_long["period"] = "long_term"

print("Short-term shape:", df_short.shape)
print("Medium-term shape:", df_medium.shape)
print("Long-term shape:", df_long.shape)
```

## 4.2 Create a Combined Dataset Without Removing Duplicates

```python
# Combine all periods without prioritizing any of them
df_tracks_all_periods = pd.concat(
    [df_short, df_medium, df_long],
    ignore_index=True
)

print("Combined dataset shape:", df_tracks_all_periods.shape)
df_tracks_all_periods.head()
```

## 4.3 Create Auxiliary Sets for Intersection Analysis

```python
# Unique track IDs per period
short_ids = set(df_short["track_id"])
medium_ids = set(df_medium["track_id"])
long_ids = set(df_long["track_id"])

# Tracks appearing in multiple periods
common_short_medium = short_ids.intersection(medium_ids)
common_short_long = short_ids.intersection(long_ids)
common_medium_long = medium_ids.intersection(long_ids)
common_all_three = short_ids.intersection(medium_ids).intersection(long_ids)

print("Tracks in all three periods:", len(common_all_three))
print("Short ∩ Medium:", len(common_short_medium))
print("Short ∩ Long:", len(common_short_long))
print("Medium ∩ Long:", len(common_medium_long))
```

# 5. Long Term Analysis

This section explores the user's **long-term listening patterns**, which represent the most stable and persistent musical preferences accumulated across several years.

The objective of this part is to understand:
- Which songs have defined the user's musical identity over time  
- Which artists appear consistently in the long-term ranking  
- Whether long-term preferences tend to be mainstream or niche  
- The dominant genres and listening patterns that form the user's musical foundation

The long-term dataset (`df_long`) contains the top 50 tracks that represent historical, consolidated musical tastes.


## 5.1 Analysis of the Top 50 Songs (Long Term)

This section analyzes the user's **Top 50 most listened tracks in the long-term window**, which reflects musical preferences accumulated over several years.  
Unlike the artist-based ranking, the track-based perspective highlights *individual songs* that have played a major role in shaping the user’s listening identity.

The objectives of this subsection are:

- To examine which songs define the user's long-term listening habits  
- To identify artists with strong presence across multiple top-ranked tracks  
- To estimate the relative importance of each track using a weighted listening metric (Estimated Listening Time, ELT)  
- To analyze how mainstream or niche the user’s core musical tastes are  
- To build per-artist summaries showing how each artist contributes to the Top 50  

This analysis provides a detailed understanding of the user's historical listening behavior at the **track level**, capturing patterns that are not visible when only considering artist-based rankings.  

The results from this part will also allow meaningful comparisons later with:
- Medium-term and short-term listening behaviour  
- Spotify’s long-term artist rankings  
- Recent emotional or contextual changes reflected in listening habits  


### 5.1.1 Top 10 Long Term Songs

```python
# Compute Estimated Listening Time (ELT)
# Formula: (51 - rank) * duration_ms
df_long["elt_raw"] = (51 - df_long["rank"]) * df_long["duration_ms"]

# Convert ELT to hours relative scale
# Explanation: elt_hours is NOT real listening time,
# but a weighted metric combining rank and duration.
df_long["elt_hours"] = df_long["elt_raw"] / (1000 * 60 * 60)

# Select Top 10 tracks by rank
long_top10 = df_long.nsmallest(10, "rank")[[
    "rank",
    "track_name",
    "artist_names",
    "duration_ms",
    "elt_hours",
    "popularity"
]]

long_top10
```

#### 5.1.1.1 Songs by Each Artist in the Top 50

```python
# Remove producers & collabs (I don't want to include)

EXCLUDED_ARTISTS = [
    "Louis Amoeba",
    "Bigla The Kid", 
    "Príncipe Palanca",
    "DAYTXNA",
    "LaBlackie"
    "DAYTXNA",
    "Kris Floyd",
    "LEÏTI",
    "J.Moods",
    "808 Blanco",
    "BEJO",
    "CeErre",
    "Jeke Mamoneos",
    "Lone",
    "N.Rizo",
    "María Reina",
    "Midas Alonso",
    "Morad",
    "Titó",
    "Alejandra Saiz",
    "Chita",
    "Selecta",
    "TALE$"
]
```

```python
# Generate a table ONLY for artists (with 2 or more) songs in the Top 50 (Long Term)

# Expand artist names
df_long_artists = df_long.copy()
df_long_artists["artist"] = df_long_artists["artist_names"].str.split(", ")
df_long_artists = df_long_artists.explode("artist")

# Filter out excluded artists (producers)
df_long_artists = df_long_artists[
    ~df_long_artists["artist"].isin(EXCLUDED_ARTISTS)
]

# Count tracks per artist
artist_track_counts_long = (
    df_long_artists["artist"]
    .value_counts()
)

# Select only artists with 2 or more tracks
artists_with_2plus_long = artist_track_counts_long[artist_track_counts_long >= 2].index.tolist()

print("Artists with 2 or more tracks in Long-Term Top 50:")
print(artists_with_2plus_long)

# Generate tables for these artists only
artist_track_tables_long = {}

for artist in artists_with_2plus_long:
    artist_df = (
        df_long_artists[df_long_artists["artist"] == artist]
        [["rank", "track_name", "artist_names", "popularity", "elt_hours"]]
        .sort_values("rank")   # Rank ascending = most listened first
        .reset_index(drop=True)
    )
    artist_track_tables_long[artist] = artist_df
    print(f"\n{artist} — Tracks in Top 50 (Long Term) ")
    display(artist_df)
```

### 5.1.2 Distribution of Artists in the Top 50 Long Term Songs

#### 5.1.2.1 Distribution of Artists in the Top 50 Long Term Songs (by Frequency)

```python
# Count artist frequency in the long-term period (filtered)

long_artist_freq = (
    df_long["artist_names"]
    .str.split(", ")
    .explode()
    .pipe(lambda s: s[~s.isin(EXCLUDED_ARTISTS)])  # Remove excluded artists
    .value_counts()
)

long_artist_freq.head(10)

```

```python
long_artist_freq_df = long_artist_freq.reset_index()
long_artist_freq_df.columns = ["artist", "count"]

print("Artist distribution in Long-Term Top Tracks (Filtered):")
long_artist_freq_df.head(20)

plt.figure(figsize=(10, 6))

sns.barplot(
    data=long_artist_freq_df.head(10),
    x="count",
    y="artist"
)

plt.title("Top Artists Appearing in Long-Term Top 50 Tracks (Filtered)")
plt.xlabel("Number of Tracks in Top 50")
plt.ylabel("Artist")
plt.show()
```

#### 5.1.2.1 Distribution of Artists in the Top 50 Long Term Songs (by ELT)

```python
# Aggregate ELT by artist
artist_elt_long = (
    df_long_artists
    .groupby("artist")["elt_hours"]
    .sum()
    .sort_values(ascending=False)
)

# Convert to DataFrame for display
artist_elt_long_df = artist_elt_long.reset_index()
artist_elt_long_df.columns = ["artist", "total_elt_hours"]

artist_elt_long_df.head(10)
```

```python
top_n = 10
plt.figure(figsize=(10, 5))

sns.barplot(
    data=artist_elt_long_df.head(top_n),
    x="total_elt_hours",
    y="artist"
)

plt.title("Top 10 Artists by Estimated Listening Time (Long Term)")
plt.xlabel("Estimated Listening Time (hours)")
plt.ylabel("Artist")
plt.show()
```

### 5.1.3 Popularity of the Top 10 Long Term Songs

```python
# Popularity statistics for the top 10 long-term tracks
pop_mean_long = long_top10["popularity"].mean()
pop_median_long = long_top10["popularity"].median()
pop_min_long = long_top10["popularity"].min()
pop_max_long = long_top10["popularity"].max()

print("Popularity Statistics — Top 10 Long-Term Tracks")
print(f"Mean popularity:   {pop_mean_long:.2f}")
print(f"Median popularity: {pop_median_long:.2f}")
print(f"Min popularity:    {pop_min_long}")
print(f"Max popularity:    {pop_max_long}")
```

## 6.1 Analysis of the Top 50 Songs (Mid Term)

new artists appear, older favourites may fade, and emotional or contextual shifts begin to surface.  
By analysing these 50 tracks carefully, we can trace the user’s progression from long-term stability to short-term evolution.
